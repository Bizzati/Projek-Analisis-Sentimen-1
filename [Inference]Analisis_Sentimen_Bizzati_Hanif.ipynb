```python
import pickle
import re
import numpy as np
```

```python
# Load model dan vectorizer
with open('tfidf_vectorizer.pkl', 'rb') as f:
    vectorizer = pickle.load(f)
with open('mlp_model.pkl', 'rb') as f:
    model = pickle.load(f)

def preprocess_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^a-z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def predict_sentiment(text):
    # Preprocessing
    preprocessed_text = preprocess_text(text)

    # TF-IDF Vectorization
    text_vector = vectorizer.transform([preprocessed_text]).toarray()

    # Predict
    prediction = model.predict(text_vector)
    predicted_class = np.argmax(prediction)

    # Map to label
    label_mapping = {0: 'negatif', 1: 'netral', 2: 'positif'}
    return label_mapping[predicted_class]

# Contoh penggunaan
test_samples = [
    "pelayanan sangat buruk dan kurir lambat",
    "aplikasi biasa saja tidak terlalu bagus",
    "pengiriman cepat dan packaging rapi"
]

for text in test_samples:
    print(f"Text: {text}")
    print(f"Prediksi: {predict_sentiment(text)}\n")
```
