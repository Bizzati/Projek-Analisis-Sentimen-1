```python
from google_play_scraper import Sort, reviews
import pandas as pd
```

# Scraping Data & Ekspor

```python
# Mengambil hingga 10.000 review dari aplikasi Paxel
result, continuation_token = reviews(
    'com.paxel',
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=10000,
    filter_score_with=None,
    # Add continuation_token to retrieve all reviews
    continuation_token=None
)

# Cek pengambilan data
if not result:
    print("No reviews found. Please check the app ID or connection.")
else:
    # Konversi ke DataFrame
    df = pd.DataFrame(result)

    print(len(result))
    print(df[['score', 'content']].head(10))
```

```python
# Simpan hasil scraping
df.to_csv('paxel_reviews.csv', index=False)
```
